```python
# Настройка логирования согласно системным инструкциям

# messge (don't commit local changes if msg=None)
msg = 'fix: for AdaptiveAvgPool3d(1) updated critic to handle temporal broadcasting and merge the Time (T) and Depth (Z) dimensions so it can be processed by standard Conv3d'

from google.colab import userdata
import subprocess

logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger(__name__)

def git_push_changes(repo_path: str, repo_name: str, branch: str = 'main', msg: str|None = None, github_user: str|None=None, github_token: str|None = userdata.get('github_token')):
    """
    Выполняет фиксацию и отправку изменений в GitHub репозиторий.

    Args:
        repo_path: Путь к локальному репозиторию в Google Drive.
        github_user: Имя пользователя GitHub.
        github_token: Personal Access Token (PAT) с правами на push.
        repo_name: Имя репозитория (например, 'And0k/3D_STA_SWGAIN').
        branch: Ветка для пуша.
    """
    try:
        # Переходим в директорию репозитория
        os.chdir(repo_path)

        # Настройка идентификации (замените на свои данные)
        subprocess.run(['git', 'config', '--global', 'user.email', userdata.get('user.email')], check=True)
        subprocess.run(['git', 'config', '--global', 'user.name', userdata.get('user.name')], check=True)

        # Добавление изменений
        subprocess.run(['git', 'add', '.'], check=True)
        logger.info("Added all changes to staging area.")

        if msg:
            # Проверка наличия изменений перед коммитом через try/except (EAFP)
            try:
                subprocess.run(['git', 'commit', '-m', msg], check=True)
                logger.info("Changes committed successfully.")
            except subprocess.CalledProcessError:
                logger.warning("No changes to commit.")
                return

        # Формирование URL с токеном для аутентификации
        remote_url = f'https://{github_user}:{github_token}@github.com/{repo_name}.git'

        # Пуш в удаленный репозиторий
        subprocess.run(['git', 'push', remote_url, branch], check=True)
        logger.info(f"Successfully pushed changes to {branch} branch.")

    except Exception as e:
        logger.exception(f"Failed to perform git push: {e}")

# Вызов функции синхронизации
git_push_changes(REPO_PATH, REPO_FULL_NAME, github_user=GH_USER, msg=msg)
```

### Добавление файла `.gitignore`


```python


def create_gitignore(repo_path: str, ignore_entries: list[str]):
    """
    Создает или обновляет файл .gitignore в указанном репозитории
    для исключения временных файлов и директорий, таких как `__pycache__`, из системы контроля версий Git.
    Args:
        repo_path: Путь к локальному репозиторию.
        ignore_entries: Список строк для добавления в .gitignore.
    """
    # Ensure the repository directory exists
    os.makedirs(repo_path, exist_ok=True)

    gitignore_path = os.path.join(repo_path, '.gitignore')
    try:
        with open(gitignore_path, 'a+') as f:
            f.seek(0) # Переходим в начало файла, чтобы прочитать текущее содержимое
            existing_content = f.read()
            for entry in ignore_entries:
                if entry not in existing_content:
                    f.write(f'\n{entry}')
                    logger.info(f"Added '{entry}' to .gitignore")
                else:
                    logger.info(f"'{entry}' already exists in .gitignore")
        logger.info(f"Successfully updated .gitignore at {gitignore_path}")
    except Exception as e:
        logger.exception(f"Failed to create or update .gitignore: {e}")

# Определите путь к вашему репозиторию
REPO_PATH = '/content/drive/MyDrive/3D_STA_SWGAIN'

# Определите общие записи для .gitignore
GITIGNORE_ENTRIES = [
    '__pycache__/',
    '*.pyc',
    '*.swp',
    '.ipynb_checkpoints/',
    '.DS_Store',
    'env/',
    'venv/',
    '*.log'
]

# Вызов функции для создания/обновления .gitignore
create_gitignore(REPO_PATH, GITIGNORE_ENTRIES)
```
